```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.master('local[1]')\
                    .appName('practice')\
                    .getOrCreate()
sc = spark.sparkContext
print(sc)
print(spark)
print(sc.appName)
```

```python
li = [1, 2, 3, 4, 5, 6]
rdd = sc.parallelize(li)
rdd.collect()
```
