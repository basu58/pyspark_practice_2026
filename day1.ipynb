```python
from pyspark import SparkContext

# 1. Initialize SparkContext (The entry point for RDDs)
sc = SparkContext("local", "RDD Example App")

# 2. Create an RDD from a list of strings
data = [
    "pyspark is great",
    "pyspark is fast",
    "learning pyspark is fun"
]
text_rdd = sc.parallelize(data)

# 3. Transformations (Lazy evaluation - nothing happens yet)
# We split lines into words, map them to (word, 1) pairs, and reduce by key
counts = text_rdd.flatMap(lambda line: line.split(" ")) \
                 .map(lambda word: (word, 1)) \
                 .reduceByKey(lambda a, b: a + b)

# 4. Action (This triggers the actual computation)
results = counts.collect()

# 5. Print results
for word, count in results:
    print(f"{word}: {count}")

# 6. Stop the context
sc.stop()
```
